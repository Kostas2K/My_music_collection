```python
import great_expectations as ge
import pandas as pd
```

```python
df=pd.read_csv('data/df_source.csv')
```

```python
df.columns
```

```python
import great_expectations as gx
from great_expectations.validator.validator import Validator
from great_expectations.execution_engine.pandas_execution_engine import PandasExecutionEngine
from great_expectations.core.batch import Batch #, BatchDefinition, BatchRequest

engine = PandasExecutionEngine()

batch = Batch(data=df)
validator = Validator(execution_engine=engine, batches=[batch])
```

```python
validator.expect_column_values_to_not_be_null(column='id')
validator.expect_column_values_to_be_unique(column='id')
validator.expect_column_distinct_values_to_contain_set(column='genres', value_set=['Rock', 'Pop', 'Jazz', 'Electronic'])
```

```python
suite=validator.build_expectation_suite()
validator.save_expectation_suite(suite)
```
